Penjelasan Program SVR
1. Membaca Dataset

    Dataset Salary_Data.csv dibaca menggunakan Pandas dan disimpan ke dalam variabel data. Dataset ini berisi informasi tentang pengalaman kerja dan gaji seseorang.

```python
import pandas as pd

# membaca dataset dan mengubahnya menjadi dataframe
data = pd.read_csv('Salary_Data.csv')
```

2. Memeriksa Informasi Dataset

    Perintah data.info() digunakan untuk melihat jumlah data, nama kolom, dan jenis datanya. Dari hasil pemeriksaan, terdapat 30 data dan tidak ditemukan nilai yang kosong.

```python
data.info()
```

3. Menampilkan Data

    Perintah data.head() digunakan untuk menampilkan lima data pertama. Tujuannya agar isi dataset bisa dilihat sebelum masuk ke proses pengolahan.

```python

data.head()
```

4. Memisahkan Atribut dan Label

    Kolom YearsExperience digunakan sebagai data pengalaman kerja, sedangkan Salary menjadi nilai gaji yang akan diprediksi. Data pengalaman kerja kemudian diubah menggunakan NumPy agar bentuknya sesuai dengan kebutuhan model SVR.

```python

import numpy as np

# memisahkan atribut dan label
X = data['YearsExperience']
y = data['Salary']
# mengubah bentuk atribut
X = np.array(X)
X = X[:,np.newaxis]
```

5. Membuat dan Melatih Model SVR

    Model dibuat menggunakan metode Support Vector Regression dengan kernel RBF dan parameter C = 1000 serta gamma = 0.05. Setelah itu, model dilatih menggunakan data yang sudah disiapkan.

```python

from sklearn.svm import SVR

# membangun model dengan parameter C, gamma, dan kernel
model = SVR(C=1000, gamma=0.05, kernel='rbf')
# melatih model dengan fungsi fit
model.fit(X,y)
```

6. Visualisasi Hasil Model

    Hasil pengolahan ditampilkan dalam bentuk grafik menggunakan Matplotlib. Titik-titik menunjukkan data asli, sedangkan garis menggambarkan hasil prediksi model berdasarkan pengalaman kerja.

```python

import matplotlib.pyplot as plt

# memvisualisasikan model
plt.scatter(X, y)
plt.plot(X, model.predict(X))
```
